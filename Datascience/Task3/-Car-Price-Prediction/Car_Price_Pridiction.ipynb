```python
# ============================================================
# OASIS INFOBYTE - DATA SCIENCE INTERNSHIP
# TASK 3 - CAR PRICE PREDICTION USING MACHINE LEARNING
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

import warnings
warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

import zipfile
import pandas as pd

zip_path = "/content/archive (1).zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    df = pd.read_csv(
        zip_ref.open("Car details v3.csv")
    )

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())

print("=" * 60)
print("CAR PRICE PREDICTION PROJECT")
print("=" * 60)

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())


# ------------------------------------------------------------
# 3. INITIAL DATA INSPECTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATA INFORMATION")
print("=" * 60)

df.info()

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ------------------------------------------------------------
# 4. REMOVE DUPLICATES
# ------------------------------------------------------------

before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)

print("\nDuplicates removed:", before_duplicates - after_duplicates)
print("Dataset shape after removing duplicates:", df.shape)


# ------------------------------------------------------------
# 5. FEATURE ENGINEERING
# ------------------------------------------------------------

# Extract car brand from the car name
df["brand"] = df["name"].str.split().str[0]

# Calculate car age
df["car_age"] = 2020 - df["year"]

# Convert mileage to numeric
df["mileage_numeric"] = (
    df["mileage"]
    .astype(str)
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)

# Convert engine capacity to numeric
df["engine_numeric"] = (
    df["engine"]
    .astype(str)
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)

# Convert max power to numeric
df["max_power_numeric"] = (
    df["max_power"]
    .astype(str)
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)

# Extract numerical torque value
df["torque_numeric"] = (
    df["torque"]
    .astype(str)
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)

print("\nFeature engineering completed!")

print("\nNew columns:")
print([
    "brand",
    "car_age",
    "mileage_numeric",
    "engine_numeric",
    "max_power_numeric",
    "torque_numeric"
])


# ------------------------------------------------------------
# 6. HANDLE MISSING VALUES
# ------------------------------------------------------------

numeric_columns = [
    "mileage_numeric",
    "engine_numeric",
    "max_power_numeric",
    "torque_numeric",
    "seats"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("\nMissing values after cleaning:")
display(df.isnull().sum())


# ------------------------------------------------------------
# 7. EXPLORATORY DATA ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 60)


# Selling price distribution
plt.figure(figsize=(10, 6))
sns.histplot(df["selling_price"], bins=50, kde=True)
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")
plt.show()


# Selling price vs car age
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="car_age",
    y="selling_price",
    alpha=0.5
)
plt.title("Selling Price vs Car Age")
plt.xlabel("Car Age (Years)")
plt.ylabel("Selling Price")
plt.show()


# Selling price vs mileage
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="mileage_numeric",
    y="selling_price",
    alpha=0.5
)
plt.title("Selling Price vs Mileage")
plt.xlabel("Mileage (km/l)")
plt.ylabel("Selling Price")
plt.show()


# Selling price by fuel type
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df,
    x="fuel",
    y="selling_price"
)
plt.title("Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Selling Price")
plt.show()


# Selling price by transmission
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=df,
    x="transmission",
    y="selling_price"
)
plt.title("Selling Price by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Selling Price")
plt.show()


# Correlation heatmap
correlation_columns = [
    "year",
    "selling_price",
    "km_driven",
    "car_age",
    "mileage_numeric",
    "engine_numeric",
    "max_power_numeric",
    "torque_numeric",
    "seats"
]

plt.figure(figsize=(12, 8))

sns.heatmap(
    df[correlation_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()


# ------------------------------------------------------------
# 8. PREPARE FEATURES AND TARGET
# ------------------------------------------------------------

features = [
    "year",
    "km_driven",
    "car_age",
    "mileage_numeric",
    "engine_numeric",
    "max_power_numeric",
    "torque_numeric",
    "seats",
    "brand",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

target = "selling_price"

X = df[features]
y = df[target]

print("\nFeatures used for prediction:")
print(features)

print("\nTarget variable:", target)


# ------------------------------------------------------------
# 9. TRAIN / TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ------------------------------------------------------------
# 10. DEFINE CATEGORICAL AND NUMERICAL FEATURES
# ------------------------------------------------------------

categorical_features = [
    "brand",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

numerical_features = [
    "year",
    "km_driven",
    "car_age",
    "mileage_numeric",
    "engine_numeric",
    "max_power_numeric",
    "torque_numeric",
    "seats"
]


# ------------------------------------------------------------
# 11. PREPROCESSING
# ------------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)


# ------------------------------------------------------------
# 12. MODEL 1 - LINEAR REGRESSION
# ------------------------------------------------------------

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)


# ------------------------------------------------------------
# 13. MODEL 2 - RANDOM FOREST REGRESSOR
# ------------------------------------------------------------

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

random_forest_model.fit(X_train, y_train)

random_forest_predictions = random_forest_model.predict(X_test)


# ------------------------------------------------------------
# 14. MODEL EVALUATION FUNCTION
# ------------------------------------------------------------

def evaluate_model(model_name, actual, predicted):

    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )

    r2 = r2_score(actual, predicted)

    print("\n" + "-" * 50)
    print(model_name)
    print("-" * 50)

    print(f"MAE  : ₹{mae:,.2f}")
    print(f"RMSE : ₹{rmse:,.2f}")
    print(f"R²   : {r2:.4f}")

    return mae, rmse, r2


# ------------------------------------------------------------
# 15. EVALUATE BOTH MODELS
# ------------------------------------------------------------

linear_mae, linear_rmse, linear_r2 = evaluate_model(
    "Linear Regression",
    y_test,
    linear_predictions
)

rf_mae, rf_rmse, rf_r2 = evaluate_model(
    "Random Forest Regressor",
    y_test,
    random_forest_predictions
)


# ------------------------------------------------------------
# 16. MODEL COMPARISON
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest Regressor"
    ],
    "MAE": [
        linear_mae,
        rf_mae
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse
    ],
    "R2 Score": [
        linear_r2,
        rf_r2
    ]
})

print("\n" + "=" * 60)
print("MODEL COMPARISON")
print("=" * 60)

display(comparison)


# ------------------------------------------------------------
# 17. IDENTIFY BEST MODEL
# ------------------------------------------------------------

if rf_r2 > linear_r2:
    best_model_name = "Random Forest Regressor"
    best_predictions = random_forest_predictions
else:
    best_model_name = "Linear Regression"
    best_predictions = linear_predictions

print("\nBest Performing Model:", best_model_name)


# ------------------------------------------------------------
# 18. ACTUAL VS PREDICTED PRICES
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.5
)

plt.xlabel("Actual Selling Price")
plt.ylabel("Predicted Selling Price")
plt.title(f"Actual vs Predicted Prices - {best_model_name}")

# Reference line
min_price = min(y_test.min(), best_predictions.min())
max_price = max(y_test.max(), best_predictions.max())

plt.plot(
    [min_price, max_price],
    [min_price, max_price],
    linestyle="--"
)

plt.show()


# ------------------------------------------------------------
# 19. RANDOM FOREST FEATURE IMPORTANCE
# ------------------------------------------------------------

rf_preprocessor = random_forest_model.named_steps["preprocessor"]
rf_model = random_forest_model.named_steps["model"]

feature_names = rf_preprocessor.get_feature_names_out()

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nTop 15 Most Important Features:")

display(feature_importance.head(15))


# Plot top 15 features
top_features = feature_importance.head(15).sort_values(
    by="Importance"
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 15 Feature Importances - Random Forest")

plt.show()


# ------------------------------------------------------------
# 20. SAMPLE PREDICTIONS
# ------------------------------------------------------------

sample_results = pd.DataFrame({
    "Actual Price": y_test.values[:10],
    "Predicted Price": best_predictions[:10]
})

print("\nSample Predictions:")

display(sample_results)


# ------------------------------------------------------------
# 21. FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)

print(f"Original dataset size : {8128:,} rows")
print(f"Final dataset size    : {len(df):,} rows")
print(f"Training samples      : {len(X_train):,}")
print(f"Testing samples       : {len(X_test):,}")

print("\nModels trained:")
print("1. Linear Regression")
print("2. Random Forest Regressor")

print("\nBest model:")
print(best_model_name)
```
